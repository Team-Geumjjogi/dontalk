````python
# json 데이터 하나로 모아 df에 저장
from pathlib import Path
import json
import pandas as pd

root = Path("../../data/raw/Training")

category_map = {
    "은행": {
        "01": "거래내역/잔액조회",
        "02": "중계요청/착오송금",
        "03": "자동이체조회",
        "04": "만기,연장/해지,수신",
        "05": "금융거래한도/비대면한도계좌",
        "06": "이자/연체금액",
        "07": "부수거래금리감면",
        "08": "대출문의(만기/연장/조회등)",
        "09": "환전문의",
    },
    "보험": {
        "01": "자동차보험상담",
        "02": "자동차사고접수",
        "03": "계약내용변경/해지",
        "04": "기타계약관련문의",
        "05": "보험금청구",
    },
    "증권": {
        "01": "HTS/MTS",
        "02": "계좌관리",
        "03": "신용거래/담보대출",
        "04": "자금이체/계좌제한",
        "05": "절세형금융상품",
        "06": "주식주문",
        "07": "증권계좌조회",
        "08": "해외주문",
    }
}

domain_map = {
    "bk": "은행",
    "ins": "보험",
    "sec": "증권",
}

# json 내부 데이터 > df 형태로 변환
def flatten_json(data):
    """JSON 내부의 중첩된 dictionary를 펼쳐서
    DataFrame 컬럼 형태로 변환"""
    result = {}

    # --------------------------------
    # source
    # --------------------------------
    # 예: source.source_institution -> source_source_institution
    if isinstance(data.get("source"), dict):
        for key, value in data["source"].items():
            result[f"source_{key}"] = value

    # --------------------------------
    # consulting
    # --------------------------------
    # 예: consulting.xxx -> consulting_xxx
    if isinstance(data.get("consulting"), dict):
        for key, value in data["consulting"].items():
            result[f"consulting_{key}"] = value

    # --------------------------------
    # qa_data
    # --------------------------------
    # qa_data는 리스트이며 첫 번째 항목 사용
    if isinstance(data.get("qa_data"), list) and data["qa_data"]:
        qa = data["qa_data"][0]

        if isinstance(qa, dict):
            for key, value in qa.items():

                # qa_data.input.xxx -> qa_data_input_xxx
                if key == "input" and isinstance(value, dict):
                    for input_key, input_value in value.items():
                        result[f"qa_data_input_{input_key}"] = input_value
                else:
                    result[f"qa_data_{key}"] = value

    return result


# 전체 json 파일 수집
all_data = []
errors = []



# Training 하위 모든 JSON 탐색
for json_path in root.rglob("*.json"):
    try:
        filename = json_path.name
        filename_without_ext = json_path.stem

        # 예: 21-1_ins_01_019557_001.json
        parts = filename_without_ext.split("_")

        if len(parts) < 3:
            raise ValueError(f"예상한 파일명 형식이 아닙니다: {filename}")

        data_field_code = parts[0]  # 예: 21-1
        domain_code = parts[1]      # 예: ins
        middle_code = parts[2]      # 예: 01

        domain_name = domain_map.get(
            domain_code,
            f"Unknown({domain_code})"
        )

        middle_name = category_map.get(
            domain_name, {}
        ).get(
            middle_code,
            f"Unknown({middle_code})"
        )

        with json_path.open("r", encoding="utf-8") as f:
            json_data = json.load(f)

        json_flattened = flatten_json(json_data)

        row = {
            "파일명": filename,
            "데이터분야별코드": data_field_code,
            "분류코드": domain_code,  # 위 코드와 컬럼명 통일
            "도메인명": domain_name,
            "중분류코드": middle_code,
            "중분류명": middle_name,
            "파일경로": str(json_path),
        }

        row.update(json_flattened)
        all_data.append(row)

    except Exception as e:
        errors.append({
            "파일경로": str(json_path),
            "오류": str(e)
        })

df = pd.DataFrame(all_data)

print(f"DataFrame 크기: {df.shape}")

df
````

````python
# Excel에서 열어 확인하기 위한 CSV 파일
# CSV용 복사본 생성: 원본 df는 변경하지 않음
df_csv = df.copy()

# Pandas 경고 없이 문자열 컬럼 선택
text_columns = df_csv.select_dtypes(
    include=["object", "string"]
).columns

# 실제 줄바꿈 및 비정상 줄 구분 문자를 문자 "\\n"으로 변환
def normalize_line_breaks(value):
    if not isinstance(value, str):
        return value

    return (
        value.replace("\r\n", "\\n")  # Windows 줄바꿈
             .replace("\n", "\\n")    # Unix 줄바꿈
             .replace("\r", "\\n")    # 옛 Mac 줄바꿈
             .replace("\u0085", "\\n")  # NEL
             .replace("\u2028", "\\n")  # LS
             .replace("\u2029", "\\n")  # PS
    )

for col in text_columns:
    df_csv[col] = df_csv[col].map(normalize_line_breaks)

# Excel 확인용 CSV 저장
df_csv.to_csv(
    "training_raw.csv",
    index=False,
    encoding="utf-8-sig",
    lineterminator="\n"
)

print(f"DataFrame 행 수: {len(df_csv):,}")
````

````python
# 데이터 정제 (마스킹 기호, 식별자 등 직관화)
from pathlib import Path
import re
import pandas as pd

# --------------------------------------------------
# 파일 경로
# --------------------------------------------------
input_parquet = Path("training_raw.parquet")

output_csv = Path("training_processed.csv")
output_parquet = Path("training_processed.parquet")

# 결측치·마스킹 표기 규칙
MISSING_VALUE = "알수없음"

# 필요하면 여기에 데이터에 등장하는 마스킹 기호를 추가하세요.
MASK_PATTERN = re.compile(r"[★▲◆●■□◇○△▼]+")

# --------------------------------------------------
# 텍스트 정제 함수
# --------------------------------------------------
def clean_text(value):
    """텍스트형 셀에만 적용하는 정제 함수"""

    if not isinstance(value, str):
        return value

    # 기존 CSV에서 줄바꿈을 문자 '\\n'으로 저장한 경우 복원
    # 이후 최종 CSV 저장 시 실제 줄바꿈은 다시 제거됨
    text = value.replace("\\n", "\n")

    # LS, PS 등 비정상 줄 구분 문자 처리
    text = (
        text.replace("\u0085", "\n")
            .replace("\u2028", "\n")
            .replace("\u2029", "\n")
    )

    # 마스킹 기호를 OOO로 통일
    text = MASK_PATTERN.sub("OOO", text)

    # 발화자 태그 변환
    # 각 줄의 시작에 있는 TX / RX만 치환
    text = re.sub(
        r"(?m)^[ \t]*TX\s*[:：]?\s*",
        "상담사: ",
        text
    )
    text = re.sub(
        r"(?m)^[ \t]*RX\s*[:：]?\s*",
        "고객: ",
        text
    )

    # 연속된 문장부호 압축: ... → ., ,,, → ,
    text = re.sub(r"([.,?!])\1+", r"\1", text)

    # 문장 종결 기호 뒤 띄어쓰기 보정
    text = re.sub(r"([.?!])([^\s])", r"\1 \2", text)

    # 줄바꿈·탭·연속 공백을 한 칸으로 압축
    text = re.sub(r"\s+", " ", text).strip()

    return text

# --------------------------------------------------
# 결측치 처리
# --------------------------------------------------
def replace_blank_with_missing(value):
    """빈 문자열 또는 공백뿐인 문자열을 알수없음으로 변경"""

    if isinstance(value, str) and not value.strip():
        return MISSING_VALUE

    return value

# --------------------------------------------------
# 원본 읽기 및 정제
# --------------------------------------------------
df_raw = pd.read_parquet(input_parquet)

# 원본을 보존하기 위해 복사본에서 작업
df_processed = df_raw.copy()

# NaN, None 처리
df_processed = df_processed.fillna(MISSING_VALUE)

# 빈 문자열/공백만 있는 값 처리
df_processed = df_processed.map(replace_blank_with_missing)

# 문자열 컬럼에 텍스트 정제 적용
text_columns = df_processed.select_dtypes(
    include=["object", "string"]
).columns

for col in text_columns:
    df_processed[col] = df_processed[col].map(clean_text)

print(f"원본 행 수: {len(df_raw):,}")
print(f"정제 후 행 수: {len(df_processed):,}")

# --------------------------------------------------
# 정제 결과 저장
# --------------------------------------------------

# 분석·재사용용: 원래 타입을 최대한 유지
df_processed.to_parquet(
    output_parquet,
    index=False
)

# Excel 확인용: 한 데이터가 CSV 한 줄만 차지하도록 저장
df_processed.to_csv(
    output_csv,
    index=False,
    encoding="utf-8-sig",
    lineterminator="\n"
)

print(f"Parquet 저장 완료: {output_parquet}")
print(f"CSV 저장 완료: {output_csv}")
````
