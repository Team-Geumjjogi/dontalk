```python
import json
from collections import defaultdict

OUTPUT_FILE = "hana_ins_faq.json"

with open(OUTPUT_FILE, "r", encoding="utf-8") as file:
    faq_data = json.load(file)

duplicate_groups = defaultdict(list)

for index, item in enumerate(faq_data):
    key = (
        item.get("question", ""),
        item.get("answer", ""),
    )
    duplicate_groups[key].append(index)

duplicates = {
    key: indexes
    for key, indexes in duplicate_groups.items()
    if len(indexes) > 1
}

duplicate_row_count = sum(
    len(indexes)
    for indexes in duplicates.values()
)

extra_duplicate_count = sum(
    len(indexes) - 1
    for indexes in duplicates.values()
)

print(f"전체 데이터: {len(faq_data)}건")
print(f"중복 그룹 수: {len(duplicates)}개")
print(f"중복에 포함된 전체 데이터: {duplicate_row_count}건")
print(f"제거 가능한 중복 데이터: {extra_duplicate_count}건")
```
